```python
median_val = df['horsepower'].median()
most_frequent = df["horsepower"].mode()[0]
df["horsepower"] = df["horsepower"].fillna(most_frequent)
median_after = df["horsepower"].median()

print(median_val)
print(median_after)
```

```python
asian = df[df["origin"].str.lower() == "asia"]
first_seven = asian[["vehicle_weight", "model_year"]].head(7)
X = first_seven.to_numpy()
XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

multiply = XTX_inv.dot(X.T).dot(y)

result = multiply.sum()
print(result)
```
